### Co2 emission (kg) per distance (km) by Aircraft Types

```python
## Using cleaned_aviation_data_v3.parquet

import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

df = pd.read_parquet("../0.Data_after_cleaning/cleaned_aviation_data_v3.parquet")
```

```python
df = df[['acft_class', 'co2_per_distance']]
df.sample(10)
df['acft_class'].unique()
```

```python
fig = px.box(
    df,
    x="acft_class",
    y="co2_per_distance",
    points="outliers",  # we can use "all", "outliers", or False
    title="CO₂ Emissions per distance (km) by Aircraft Class",
    labels={"co2": "CO₂ Emissions (kg)", "acft_class": "Aircraft Class"},
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Aircraft Class",
    yaxis_title="CO₂ Emissions (kg) per distance (km)",
    margin=dict(l=60, r=60, t=60, b=60),
)

fig.update_xaxes(tickangle=45)

fig.show()

### Should not be UNKNOWN !!!
```

**Average Co2 emission (kg) per distance (km) by aircraft types**

```python

avg_co2 = df.groupby('acft_class', as_index=False)['co2_per_distance'].mean()

fig = px.bar(
    avg_co2,
    x="acft_class",
    y="co2_per_distance",
    title="Average CO₂ Emissions per distance (kg/km) by Aircraft Class",
    labels={"co2_per_distance": "Average CO₂ Emissions (kg/km)", "acft_class": "Aircraft Class"},
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Aircraft Class",
    yaxis_title="Average CO₂ Emissions (kg/km)",
    margin=dict(l=60, r=60, t=60, b=60),
)

fig.update_xaxes(tickangle=45)

fig.show()
```

```python
import plotly.express as px

# 1. Group and calculate the mean for both columns
avg_stats = df.groupby('acft_class', as_index=False)[['co2_per_distance', 'distance_km']].mean()

# 2. Round for prettier printing
avg_stats['co2_per_distance'] = avg_stats['co2_per_distance'].round(4) # 4 decimals for emissions
avg_stats['distance_km'] = avg_stats['distance_km'].round(2)           # 2 decimals for distance

# 3. Print the averages
print("Average Distance and CO₂ Emissions per Distance by Aircraft Class:")
print(avg_stats)
```
